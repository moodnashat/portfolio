# 🐍 Sales Exploratory Data Analysis
Portfolio project using Pandas and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../excel-sales-analysis/data/sales_data.csv', parse_dates=['Order_Date'])
df['Sales'] = df['Quantity'] * df['Unit_Price']
df['Month'] = df['Order_Date'].dt.to_period('M').astype(str)
df.head()
```

## Data Quality

```python
df.info()
df.isna().sum()
```

## KPIs

```python
kpis = {'Total Sales': df['Sales'].sum(), 'Total Orders': df['Order_ID'].nunique(), 'Total Quantity': df['Quantity'].sum(), 'Average Order Value': df['Sales'].mean()}
pd.Series(kpis)
```

## Regional and Category Analysis

```python
display(df.groupby('Region')['Sales'].sum().sort_values(ascending=False))
display(df.groupby('Category')['Sales'].sum().sort_values(ascending=False))
```

## Monthly Trend

```python
monthly = df.groupby('Month', as_index=False)['Sales'].sum()
monthly.plot(x='Month', y='Sales', kind='line', marker='o', legend=False, figsize=(8,5))
plt.title('Monthly Sales Trend')
plt.ylabel('Sales')
plt.show()
```

## Key Findings
- Total sales: 450,350.
- 24 orders and 142 units were recorded.
- Cairo is the leading region by sales.
- Electronics is the leading category.
- Laptop is the highest-revenue product.
- April is the strongest month in this dataset.
